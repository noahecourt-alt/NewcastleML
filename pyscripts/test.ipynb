```python
from pathlib import Path
import gzip
import json
import shutil
import pandas as pd

# Works whether the notebook kernel starts in pyscripts/ or the project root.
ROOT = Path.cwd()
if not (ROOT / "MAMA_MIA").is_dir():
    ROOT = ROOT.parent

SOURCE = ROOT / "MAMA_MIA"
OUTPUT = ROOT / "MAMA_MIA_formatted_all"
COHORTS = {"DUKE", "ISPY1", "ISPY2"}  # NACT intentionally excluded

IMAGE_ROOT = SOURCE / "images"
MASK_ROOT = SOURCE / "segmentations" / "expert"

def phase_index(path):
    return int(path.name.removesuffix(".nii.gz").rsplit("_", 1)[1])

def decompress_copy(source, destination):
    """Create destination only if it does not already exist."""
    if destination.exists():
        return

    destination.parent.mkdir(parents=True, exist_ok=True)
    temporary = destination.with_suffix(destination.suffix + ".part")

    with gzip.open(source, "rb") as source_file:
        with temporary.open("wb") as destination_file:
            shutil.copyfileobj(source_file, destination_file, length=1024 * 1024)

    temporary.replace(destination)

clinical_rows = []
report_rows = []

for image_folder in sorted(IMAGE_ROOT.iterdir()):
    if not image_folder.is_dir():
        continue

    patient_id = image_folder.name.upper()
    cohort = patient_id.split("_", 1)[0]

    if cohort not in COHORTS:
        continue

    phase_files = sorted(image_folder.glob("*.nii.gz"), key=phase_index)
    mask_source = MASK_ROOT / f"{patient_id.lower()}.nii.gz"
    info_source = SOURCE / "patient_info_files" / f"{patient_id.lower()}.json"

    status = "included"

    if not phase_files:
        status = "skipped_no_images"
    elif not mask_source.exists():
        status = "skipped_no_expert_mask"
    elif not info_source.exists():
        status = "skipped_no_metadata"
    else:
        with info_source.open(encoding="utf-8") as f:
            metadata = json.load(f)

        pcr = metadata.get("primary_lesion", {}).get("pcr")
        if pcr not in (0, 1):
            status = "skipped_missing_pcr"

    report_rows.append(
        {
            "patient_id": patient_id,
            "cohort": cohort,
            "n_phases": len(phase_files),
            "status": status,
        }
    )

    if status != "included":
        continue

    target_patient_folder = OUTPUT / f"{patient_id}_data"

    for phase_file in phase_files:
        phase = phase_index(phase_file)
        decompress_copy(
            phase_file,
            target_patient_folder / f"{patient_id}_{phase:04d}.nii",
        )

    decompress_copy(mask_source, target_patient_folder / "mask.nii")

    clinical_rows.append(
        {
            "patient_id": patient_id,
            "pcr": int(pcr),
            # Phase index only; not real DCE acquisition times.
            "acquisition_times": str(list(range(len(phase_files)))),
            "time_axis": "phase_index",
            "n_phases": len(phase_files),
            "source_dataset": cohort,
            "segmentation": "expert",
        }
    )

OUTPUT.mkdir(exist_ok=True)

pd.DataFrame(report_rows).to_csv(OUTPUT / "format_report.csv", index=False)

clinical = pd.DataFrame(clinical_rows).set_index("patient_id")
clinical.to_excel(OUTPUT / "clinical_and_imaging_info.xlsx")

print(f"Formatted patients: {len(clinical)}")
print(f"Output: {OUTPUT}")
print(pd.DataFrame(report_rows).groupby(["cohort", "status"]).size())
```
